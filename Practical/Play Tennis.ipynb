```python
ximport pandas as pd
from sklearn.naive_bayes import CategoricalNB
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.metrics import confusion_matrix
```

```python
df = pd.read_csv('/content/play_tennis.csv')
df
```

```python
le = {}
df_enc = df.copy()
for col in df_enc.columns:
    le[col] = LabelEncoder()
    df_enc[col] = le[col].fit_transform(df_enc[col])
df_enc
```

```python
x = df_enc[['day','outlook','temp','humidity','wind',]]
y = df_enc['play']
```

```python
x
```

```python
y
```

```python
model = CategoricalNB()
model.fit(x,y)
```

```python
test =[['D1','Sunny','Hot','High','Weak']]
```

```python
test
```

```python
print("\n---- Prediction ----------------------------------")
for t in test:
  row = [[
      le['day'].transform([t[0]])[0],
      le['outlook'].transform([t[1]])[0],
      le['temp'].transform([t[2]])[0],
      le['humidity'].transform([t[3]])[0],
      le['wind'].transform([t[4]])[0]
  ]]
  pre = le['play'].inverse_transform(model.predict(row))[0]
  prob = model.predict_proba(row)[0]
  print(f"Input :day={t[0]} outlook={t[1]} temp={t[2]} humidity={t[3]} wind={t[4]}")
  print(f"Output: play={pre}")
  print(f"Result : play -> {pre}")
  print(f"P(No) = {prob[0]:.4f}, P(Yes) = {prob[1]:.4f}")
  print("--------------------------------------------------")
```

```python
from sklearn.metrics import accuracy_score

y_pred = model.predict(x)
print(f"\n Training Accuracy: {accuracy_score(y,y_pred)*100:.2f}%")
```
